```python
def test_planets(position_now, time_now, step_size, total_distance_moved, number_minshara_hit):

    # 1. is step_size an integer?
    if not float(step_size).is_integer():
        raise Exception ("Step size must be a whole number.")

    # 1.5 calculate distance
    total_distance_moved += np.sqrt(step_size**2 + 1)

    # 2. does it intersect with any planet?
    planet_locations = np.loadtxt("/Users/tannazazad/TopicsInSciComp/Notebooks/data/planet_locations.dat")
    closest_planets_time = planet_locations[np.equal(time_now, planet_locations[:,0])] # get planet(s?) with same time value

    for closest_planet in closest_planets_time: # handles if multiple planets at same t-value
        
        closest_planet_location = closest_planet[1] # can ignore t-value (closest_planet[0])
        closest_planet_type = closest_planet[2]
        
        if position_now == closest_planet_location: # 2.1. does it hit a planet?
            assert closest_planet_type == 1, "Ship hit a D-class planet, try again." # 2.2. does it hit a D-class planet?
            number_minshara_hit += 1

    if time_now == 15:
        # 3. does it return to y = 0 at the end?
        assert position_now==0, "Ship must return to y=0 after flight."

        # 4. has it hit all minshara class planets?
        total_number_minshara_planets = planet_locations[:, 2].sum()
        assert number_minshara_hit == total_number_minshara_planets, "Must hit all M-class planets."

    return total_distance_moved, number_minshara_hit
```
